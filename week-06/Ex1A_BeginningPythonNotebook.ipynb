## Typecasting

In each of the following cells, cast the value into the specified datatype, and run the cell to check your results:

```python
# Transform the below value into an integer using the int() function
number = int('12')
print(number)
print(type(number))
```

```python
# Transform the below value into a float using the float() function
number = float(55)
print(number)
print(type(number))
```

```python
# Transform the below value into a string using the str() function
number_text = str(765)
print(number_text)
print(type(number_text))
```

```python
# Transform the below value into a boolean using the bool() function
value = bool(0)
print(value)
print(type(value))
```

```python
# What other number or string value(s) can be typecast as boolean values?
print("bool(1):", bool(1))
print("bool(-3):", bool(-3))
print("bool(0.0):", bool(0.0))
print("bool('hello'):", bool('hello'))
print("bool(''):", bool(''))
```

```python
# What's the difference between 0 and 0.0? Write your answer below as a code comment.
# 0 is an integer (int), while 0.0 is a floating-point number (float).
print(type(0))
print(type(0.0))
```
