# Exercise 2.A Using built-in functions

## Lab 2. Parsing strings – part codes

```python
def get_supplier_code(part_code):
    # Take every character before the colon.
    return part_code[:part_code.index(':')]
```

```python
def get_product_num(part_code):
    # Take the characters between the colon and the dash.
    colon = part_code.index(':')
    dash = part_code.index('-')
    return part_code[colon + 1:dash]
```

```python
def get_size(part_code):
    # Take every character after the dash.
    return part_code[part_code.index('-') + 1:]
```

```python
part_code_1 = 'ACME:123-L'
part_code_2 = 'DI:12345-M'
part_code_3 = 'ACE:1-12'
```

```python
supplier = get_supplier_code(part_code_1)
product_num = get_product_num(part_code_1)
size = get_size(part_code_1)
print(f"For {part_code_1}, the supplier is {supplier}. Part number {product_num} is available in size {size}.")
```

```python
supplier = get_supplier_code(part_code_2)
product_num = get_product_num(part_code_2)
size = get_size(part_code_2)
print(f"For {part_code_2}, the supplier is {supplier}. Part number {product_num} is available in size {size}.")
```

```python
supplier = get_supplier_code(part_code_3)
product_num = get_product_num(part_code_3)
size = get_size(part_code_3)
print(f"For {part_code_3}, the supplier is {supplier}. Part number {product_num} is available in size {size}.")
```

### Optional bonus: Part codes in a list

```python
part_codes = [part_code_1, part_code_2, part_code_3]
for part_code in part_codes:
    print(
        f"For {part_code}, the supplier is {get_supplier_code(part_code)}. "
        f"Part number {get_product_num(part_code)} is available in size {get_size(part_code)}."
    )
```
