## Lab 3. Employee data

This lab exercise makes updates to information extracted from an employee database.

```python
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
```

### Update 1: Mark Blick → Mark Blick-Hawley

```python
mark = hr_list[1]
hr_list[1] = (mark[0], mark[1], 'Mark Blick-Hawley', mark[3])
```

```python
print(hr_list)
```

### Update 2: Jim Smith moves from OPS to CS, salary $60,000

```python
jim = hr_list[3]
hr_list[3] = (jim[0], 'CS', jim[2], '60000')
```

```python
print(hr_list)
```

### Display updated employee records

```python
for employee_id, dept_code, name, salary in hr_list:
    print(f"{employee_id} | {dept_code} | {name} | {int(salary):,}")
```
