# Exercise 2.C — Lab 1: Defining custom functions

```python
def favorite_things(name, movie, music):
    return f"Name: {name}\nFavorite movie: {movie}\nFavorite musician/band: {music}"

print(favorite_things("Ricardo", "Harry Potter", "Mac Miller"))
```

```python
def why_im_here(purpose):
    return purpose

print(why_im_here("I joined Year Up United to learn data analytics and prepare for a career in cloud data."))
```

```python
def favorite_place(place, why):
    return f"One of my favorite places is {place} because {why}."
```

```python
print(favorite_place("Los Angeles", "I enjoy the music and entertainment scene"))
```

```python
print(favorite_place("Santa Monica", "I enjoy the ocean views"))
```
