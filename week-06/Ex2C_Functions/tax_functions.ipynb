# Exercise 2.C — Lab 3: Tax functions

Define functions for Social Security, Medicare, and federal withholding taxes, and test each function with three paychecks.

```python
def get_soc_sec_tax(gross_pay):
    return gross_pay * 0.062
```

```python
def get_medicare_tax(gross_pay):
    return gross_pay * 0.0145
```

```python
def get_federal_tax(gross_pay, withholding_code):
    if withholding_code == 0:
        rate = 0.23
    elif withholding_code == 1:
        rate = 0.21
    elif withholding_code == 2:
        rate = 0.195
    elif withholding_code == 3:
        rate = 0.185
    else:
        rate = 0.18 - (withholding_code - 4) * 0.005
    return gross_pay * rate
```

```python
print("Person 1: Gross pay $750.00, withholding code 0")
print(f"Social Security tax: ${get_soc_sec_tax(750):.2f}")
print(f"Medicare tax: ${get_medicare_tax(750):.2f}")
print(f"Federal tax: ${get_federal_tax(750, 0):.2f}")
```

```python
print("Person 2: Gross pay $1,550.00, withholding code 2")
print(f"Social Security tax: ${get_soc_sec_tax(1550):.2f}")
print(f"Medicare tax: ${get_medicare_tax(1550):.2f}")
print(f"Federal tax: ${get_federal_tax(1550, 2):.2f}")
```
