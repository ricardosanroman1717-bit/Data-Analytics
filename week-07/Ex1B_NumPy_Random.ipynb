## Generating Random Choices from an Array

```python
students = np.array([
    'Aalexus White',
    'Ajani Lewis',
    'Alvie Ahmed',
    'Amaiyah Young',
    'Arber Xharra',
    'Ardita Mrapshtija',
    'Arionsae Mccutcheon',
    'Brandon Silverio',
    'Brianna Farrell',
    'Christopher Springer',
    'Denae Mendez',
    'Djeneba Sacko',
])
print(students)
```

```python
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
# This is not a fair way to make groups because choice can pick a student more than once and leave others out.
```

```python
# Probabilities in the same order as the students array:
# Aalexus 30%, Ajani 0%, Alvie 10%, Amaiyah 10%, Arber 8%, Ardita 7%.
# Arionsae 7%, Brandon 6%, Brianna 6%, Christopher 6%, Denae 5%, Djeneba 5%.
# They total 100%; Aalexus is most likely and Ajani cannot be selected.
probabilities = [0.30, 0.00, 0.10, 0.10, 0.08, 0.07, 0.07, 0.06, 0.06, 0.06, 0.05, 0.05]
call_on_weighted = np.random.choice(students, size=5, p=probabilities)
print(call_on_weighted)
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
