## Sample Arrays

```python
r_1 = np.random.randint(1, 21, size=2)
r_2 = np.random.randint(1, 21, size=(1, 2))
r_3 = np.random.rand(2, 1)
r_4 = np.random.randint(20, 101, size=(4, 5))
r_5 = np.random.rand(4, 5)
print(r_1, r_2, r_3, r_4, r_5, sep='\n')
```

## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

The shuffle moved whole rows around rather than mixing up the values inside each row, because the array is two-dimensional.

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
```

I would use `shuffle()` when I want to change the original array. I would use `permutation()` when I want a shuffled copy without changing the original.

## Mathematical Operations

```python
print(r_1, r_1.shape)
print(r_2, r_2.shape)
added_1 = np.add(r_1, r_2)
print(added_1)
print(added_1.shape)
```

```python
print(r_1, r_1.shape)
print(r_3, r_3.shape)
added_2 = np.add(r_1, r_3)
print(added_2)
print(added_2.shape)
```

```python
print(r_2, r_2.shape)
print(r_4, r_4.shape)
try:
    print(np.add(r_2, r_4))
except ValueError as error:
    print(type(error).__name__ + ':', error)
# Broadcasting fails because shapes (1, 2) and (4, 5) cannot be matched from the right.
```

```python
print(np.add(r_4, np.array([1])))
```

```python
print(np.add(r_4, np.array([[1]])))
```

```python
print(np.add(r_4, np.array([1, 1, 1, 1, 1])))
```

```python
try:
    print(np.add(r_4, np.array([[1], [1]])))
except ValueError as error:
    print(type(error).__name__ + ':', error)
# The shapes (4, 5) and (2, 1) cannot broadcast because 4 and 2 do not match.
```

Broadcasting lets NumPy add arrays of different shapes by repeating values when a dimension is 1. The `add` function adds matching positions, or repeats a smaller row/column/scalar across the larger array. The result shape is determined by the compatible dimensions of both inputs; if corresponding dimensions differ and neither is 1, it fails.

## Grocery Problem

```python
# Category order: vegetables, fruits, boxes of snacks, beverages.
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
print(bag1)
print(bag2)
```

```python
print(f'The first bag contains {np.sum(bag1)} items, and the second bag contains {np.sum(bag2)} items.')
```

```python
categories = np.add(bag1, bag2)
print(f'There are {categories[0]} vegetables, {categories[1]} fruits, {categories[2]} boxes of snacks, and {categories[3]} beverages in the two grocery bags.')
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])
```

```python
# Ben missed Monday, so that day is recorded as 0 doors.
ben = np.array([0, 5, 5, 3, 4])
```

```python
# Carla missed Wednesday, so that day is recorded as 0 doors.
carla = np.array([3, 6, 0, 3, 4])
```

```python
all_knocks = np.array([ali, ben, carla])
print(all_knocks)
```

```python
totals_by_person = np.sum(all_knocks, axis=1)
print(f'Ali: {totals_by_person[0]}, Ben: {totals_by_person[1]}, Carla: {totals_by_person[2]}')
print(f'Group total: {np.sum(all_knocks)}')
```

```python
daily_knocks = np.sum(all_knocks, axis=0)
estimated_votes = np.cumsum(daily_knocks) * 0.5
print(estimated_votes)
```
