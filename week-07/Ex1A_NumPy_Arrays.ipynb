# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop prints a whole row of doubled numbers at a time. The nested loop goes into each row and prints each doubled number on its own line.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
print(loop_result)
# x is a whole row, not an individual number; loop_result is overwritten until only the last doubled row remains.
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
try:
    stacked_array_2 = np.stack((joined_array_1, joined_array_2))
    print(stacked_array_2)
except ValueError as error:
    print(type(error).__name__ + ':', error)
# The arrays cannot be stacked: one is (4, 3) and the other is (2, 6); stack needs matching shapes.
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
try:
    another_2 = np.concatenate((joined_array_1, arrA), axis=1)
    print(another_2)
except ValueError as error:
    print(type(error).__name__ + ':', error)
# Joining across columns fails because joined_array_1 has 4 rows, but arrA only has 2.
```

```python
horizontal = np.hstack((arrA, arrA))  # Adds columns side by side.
vertical = np.vstack((arrA, arrA))    # Adds rows underneath each other.
depth = np.dstack((arrA, arrA))       # Adds a new depth dimension.
print(horizontal)
print(vertical)
print(depth)
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)
# There are 2 arrays, each still 2-dimensional (2 rows by 3 columns).
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# There are 3 arrays, each 2-dimensional: the first is (2, 3), and the other two are (1, 3).
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7, 3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
